```python
from langchain_core.documents import Document
from langchain_openai import OpenAIEmbeddings

documents = [
    Document(page_content="Dogs are great companions, known for their loyalty and friendliness.", 
             metadata={"source":"mammal-pets-doc"}),
    Document(page_content="Cats are independent pets that often enjoy their own space.", 
             metadata={"source":"mammal-pets-doc"}),
]

embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
embed_query = embeddings.embed_documents([d.page_content for d in documents])
print(f"vectors length {len(embed_query[0])}")
```

```python
# vector store => (1) InMemoryVectorStore

from langchain_core.vectorstores import InMemoryVectorStore

vector_store = InMemoryVectorStore(embeddings)

vector_store.add_documents(documents=documents)

results = vector_store.similarity_search_with_score("which pets prefer to be alone?", k=1)

for doc, score in results:
    print(f"{score:.3f} : {doc.page_content}")
```

#### Text Splitter (텍스트 분할)
- LLM의 Context Window 의 제한이 있어 문서 전체를 한 번에 넣을 수 없다.
- Text Splitter 는 문서를 적절한 크기의 청크(chunk)로 분할
- chunk_size: 청크 최대 크기 
- chunk_overlap: 청크 간 겹치는 부분(문맥 유지)
- RecursiveCharacterTextSplitter 가 대부분 좋은 성능을 냄

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 1. 문서로드

loader = PyPDFLoader("./data/제주관광가이드.pdf")
docs = loader.load()

# 2. 텍스트 분할
splitter = RecursiveCharacterTextSplitter(separators=["\n\n","\n", " "],
                                        chunk_size=500, chunk_overlap = 50, length_function=len)

chunks = splitter.split_documents(docs)

print(f"원본 페이지 수 {len(docs)}")
print(f"분할된 청크 수 {len(chunks)}")
print(f"첫 청크 길이 {len(chunks[0].page_content)}")
print(f"첫 청크 내용 {len(chunks[0].page_content)}")
print(f"첫 청크 메타데이터 {len(chunks[0].metadata)}")
```

```python
text = "파이썬은 1991년에 발표된 언어입니다. " * 20

configs = [
    {"chunk_size":100, "chunk_overlap":0},
    {"chunk_size":100, "chunk_overlap":20},
    {"chunk_size":200, "chunk_overlap":50},
]

for cfg in configs:
    sp = RecursiveCharacterTextSplitter(**cfg)
    chunks = sp.split_text(text)
    print(f"size={cfg['chunk_size']}, overlap={cfg['chunk_overlap']}")
    print(f" -> {len(chunks)} 개 청크, 첫 청크  ={chunks[0][:30]}")
    if len(chunks) > 1:
        print(f"첫 청크 끝 : ", {chunks[0][-20:]})
        print(f"둘째 청크 시작 : ", {chunks[1][:20]})
```

#### Vector Store

```python
from langchain_chroma import Chroma

# 1. 문서로드
loader = PyPDFLoader("./data/제주관광가이드.pdf")
docs = loader.load()

# 2. 청킹 - 텍스트 분할
splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap = 50)
chunks = splitter.split_documents(docs)
print(f"청크 수 {len(chunks)}")

# 3. 임베딩
openai_embeddings = OpenAIEmbeddings(model="text-embedding-3-small")

# 4. vector store에 저장
vector_store = Chroma.from_documents(documents=chunks, 
                                     embedding=openai_embeddings, 
                                     persist_directory="./db/chroma_db", 
                                     collection_name="jeju"
                                     )

print("ChromaDB 저장 완료")

# 5. 유사도 검색
results = vector_store.similarity_search("제주 7대 관광명소는?", k=3)
for i, doc in enumerate(results):
    print(f"\n[결과 {i}] 출처 : {doc.metadata}")
    print(f"{doc.page_content[:150]}")
```

```python
# FAISS

from langchain_community.vectorstores import FAISS

faiss_store = FAISS.from_documents(documents=chunks, embedding=openai_embeddings)

# 5. 유사도 검색
results = faiss_store.similarity_search("제주 7대 관광명소는?", k=3)
for i, doc in enumerate(results):
    print(f"\n[결과 {i}] 출처 : {doc.metadata}")
    print(f"{doc.page_content[:150]}")


faiss_store.save_local("./db/faiss/jeju_index")
print("FAISS 저장 완료")
```

```python
from langchain_chroma import Chroma
from langchain_core.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

# 1. 문서로드
loader = PyPDFLoader("./data/Summary of ChatGPTGPT-4 Research.pdf")
docs = loader.load()

# 2. 청킹 - 텍스트 분할
splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap = 50)
chunks = splitter.split_documents(docs)
print(f"청크 수 {len(chunks)}")

# 3. 임베딩
openai_embeddings = OpenAIEmbeddings(model="text-embedding-3-small")

# 4. vector store에 저장
vector_store = Chroma.from_documents(documents=chunks, 
                                     embedding=openai_embeddings, 
                                     persist_directory="./db/chroma_db", 
                                     collection_name="research",
                                     ids=[f"research-{i}" for i in range(len(chunks))]
                                     )

print("ChromaDB 저장 완료")

# 5. RAG(Retriever) : 벡터스토어에 내용을 검색 가능한 인터페이스로 변경
retriever = vector_store.as_retriever(search_type="similarity", search_kwargs={"k":4})

prompt = ChatPromptTemplate.from_messages([
    ('system','다음 컨텍스트를 참고하여 질문에 답하세요\n컨텍스트에 없는 내용은 모른다고 답하세요\n\n컨텍스트 : {context}'),
    ('human', '{question}')
])

llm = init_chat_model(model="gpt-6-luna", reasoning={"effort":"low"})

def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

rag_chain = ({
    'context':retriever | format_docs, 
    'question':RunnablePassthrough()
    } | prompt | llm | StrOutputParser()
)


answer = rag_chain.invoke("where can i use chatGPT?")
print(answer)
```

```python
from langchain_core.runnables import RunnableParallel, RunnablePassthrough

def format_docs_with_source(docs):
    result =[]
    for doc in docs:
        src = doc.metadata.get('source','알 수 없음')
        page = doc.metadata.get('page','?')
        result.append(f"[출처] : {src}, p.{page}\n{doc.page_content}")
    return "\n\n".join(result)

rag_with_source = RunnableParallel(answer=rag_chain, sources=retriever)
result = rag_with_source.invoke("where can i use chatGPT?")

print("=== 답변 ===")
print(result['answer'])
print("=== 출처 ===")
for doc in result['sources']:
    print(f"- {doc.metadata.get('source')} p.{doc.metadata.get('page')}")
    
```

```python
# gpu : bge-m3 설치되어 있음

from langchain_chroma import Chroma
from langchain_core.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
from langchain_ollama import OllamaEmbeddings

# 1. 문서로드 및 분할
loader = PyPDFLoader("./data/2026 상 삼성SDS 직무기술서.pdf")
docs = loader.load()

# 2. 청킹 - 텍스트 분할
splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap = 50)
chunks = splitter.split_documents(docs)
print(f"청크 수 {len(chunks)}")

# 3. 임베딩 / vector store에 저장
ollama_embeddings = OllamaEmbeddings(model="bge-m3")

def create_vectorstore(chunks, embedding, collection_name):
    return  Chroma.from_documents(documents=chunks, 
                                        embedding=embedding, 
                                        persist_directory="./db/chroma_db", 
                                        collection_name=collection_name
                                        )

ollama_vectorstore = create_vectorstore(chunks, ollama_embeddings, collection_name="ollama_SamsungSDS")
openai_vectorstore = create_vectorstore(chunks, openai_embeddings, collection_name="openai_SamsungSDS")
google_vectorstore = create_vectorstore(chunks, google_embeddings, collection_name="google_SamsungSDS")

# 5. RAG(Retriever) : 벡터스토어에 내용을 검색 가능한 인터페이스로 변경
def create_retriever(vector_store):
    return vector_store.as_retriever(search_type="similarity", search_kwargs={"k":4})

ollama_retriever = create_retriever(ollama_vectorstore)
openai_retriever = create_retriever(openai_vectorstore)
google_retriever = create_retriever(google_vectorstore)


def print_retrieved_docs(title, retriever, query):
    docs = retriever.invoke(query)

    print("\n"+"="*50)
    print(title)
    print("="*50)

    for i, doc in enumerate(docs):
        print(f"page={doc.metadata.get("page")}")
        print(doc.page_content)

query = "필요한 역량 이외의 준비할건 뭐가있을까?"

print_retrieved_docs("Ollama Embedding Result", ollama_retriever, query)
print_retrieved_docs("Openai Embedding Result", openai_retriever, query)
print_retrieved_docs("Google Embedding Result", google_retriever, query)
```

### OCR
- PDF가 이미지라면 시도

```python
import pymupdf

pdf = pymupdf.open("./data/2026 상 삼성전자 DX부문 직무기술서.pdf")

# pdf 페이지 => 이미지 저장
for page_num in range(len(pdf)):
    page = pdf[page_num]
    pix = page.get_pixmap(dpi=300)
    pix.save(f"./ocr/page_{page_num:02d}.png")
```

```python
# 저장된 이미지에서 텍스트 추출하기

import easyocr

reader = easyocr.Reader(['ko','en'])
result = reader.readtext("./ocr/page_29.png", detail=0, paragraph=True)
print("\n".join(result))
```

```python
import json

pages = []

for page_num in range(31):
    image_path = f"./ocr/page_{page_num:02d}.png"

    result = reader.readtext(image_path, detail=0, paragraph=True)
    text = "\n".join(result)

    pages.append({"page":page_num+1, "content":text})

with open("./data/samsung_dx_orc.json", "w", encoding = "utf-8")as f:
    json.dump(pages, f, ensure_ascii=False, indent=2)
```

```python
# json 읽어오기 => Document => 분할하기

from langchain_core.documents import Document

with open("./data/samsung_dx_orc.json", "r", encoding = "utf-8")as f:
    pages = json.load(f)

docs = []

for page in pages:
    docs.append(
        Document(
            page_content=page['content'],
            metadata = {"page":page['page'], "source":"samsung_dx_ocr"}
        )
    )
```
