```python
# temperature = 0 : 확률이 높은 토큰을 선택
# temperature = 1.5 : 다양하게 낮은 토큰도 선택할 수 있음(실험적인 텍스트 생성)

def ask(messages ,think=False , temperature=0):
    response = client.chat(
    model=MODEL,
    messages=messages ,
    stream = False,
    think = think,
    options= {"temperature":temperature}  
    )

    return response.message.content
```

```python
messages=[{ "role":"user","content": "나 지금 몹시 매우 간절하게 배고파"}]

ask(messages=messages, temperature=1.5)
```

```python
response = client.chat(
    model=MODEL,
    messages=messages,
    stream = True,
    think = False,
    )

for chunk in response:
    print(chunk['message']['content'], end="", flush=True)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv("../src/llm/data/reviews.csv")

# 학습/테스트 데이터 분리
text_train, text_test, y_train, y_test = train_test_split(df['text'], df['label'], test_size=0.2, random_state=42, stratify=df['label'])

SYSTEM = "당신은 영화리뷰 감정 분석가이다. 리뷰를 읽고, '긍정' 또는 '부정' 한 단어로만 답한다."

def classify(review, shots=None):
    messages = [{"role":"system", "content":SYSTEM}]

    for text, label in (shots or []):
        messages.append({"role":"user", "content":f"리뷰 {text}"})
        messages.append({"role":"assistant", "content":f"{label}"})
    messages.append({"role":"user", "content":f"리뷰{review}"})

    answer= ask(messages)

    if "긍정" in answer: return 1
    if "부정" in answer: return 0

    return -1
```

```python
sample, y_sample = text_test[:100], y_test[:100]
preds = [classify(t) for t in sample]
acc = (pd.Series(preds, index=y_sample.index) == y_sample).mean()
print("제로샷 정확도:", round(acc, 3), "/ 형식 어긴 답 : ", preds.count(-1))
```

```python
shots = [
    ("배우들 연기가 살아있다. 강추", "긍정"),
    ("시간 아깝다 진짜", "부정"),
    ("ost만 기억에 남는 졸작", "부정"),
    ("두번 봐도 또 울었다", "긍정"),
]

preds_few = [classify(t, shots) for t in sample]
acc = (pd.Series(preds_few, index=y_sample.index) == y_sample).mean()
print("퓨삿 정확도:", round(acc, 3), "/ 형식 어긴 답 : ", preds_few.count(-1))
```

#### 출력형식 json, pydantic

```python
import json

schema = {
    "type":"object",
    "properties":{
        "label":{"type":"string", "enum":['긍정','부정']},
        "reason":{"type":"string"}
    },
    "required":["label","reason"]
}

r = client.generate(
    model=MODEL,
    prompt=f"{SYSTEM}\n이유도 한 문장으로 적어줘\n리뷰:연기가 좋고 정말 재밌어요.",
    stream = False,
    think = False,
    format=schema,
    options={"temperature":0}
)
print(r.response)
result = json.loads(r.response)
```

```python
print(result['label'], result['reason'])
```

```python
from pydantic import BaseModel, Field

class SentimentResult(BaseModel):
    sentiment:str
    score:float = Field(ge=0.0, le=1.0)

prompt="""
다음 문장의 감정을 분석하라.

문장:
오늘 정말 행복한 하루야.
"""
result = client.chat(model=MODEL, messages=[{"role":"user", "content":prompt}], format=SentimentResult.model_json_schema())
print(result)

# 답변 형식 점검
response = SentimentResult.model_validate_json(result.message.content)
print(response.sentiment)
print(response.score)
```

```python
# 뉴스 분석 : summary(str), sentiment(positive, negative, neutral), keyword:[과학, 기술, ..]
from typing import Literal


class NewsResult(BaseModel):
    summary:str
    sentiment:Literal['positive', 'negative', 'neutral']
    keyword:list[str]

prompt="""
다음 뉴스를 분석하라.
단, format으로 주어진 구조로 출력해라.

기사:
 장이 열리자마자 코스피가 5,000선 아래로 급락하면서 개인투자자(개미)들은 그야말로 패닉에 빠졌다.

특히 국내 증시 '투톱'이자 이른바 '국민주'로 불리는 삼성전자와 SK하이닉스가 10% 안팎의 하락률을 보이면서 이제는 버티기조차 어렵다는 하소연이 쏟아지고 있다.

6일 코스피는 8% 넘게 급락하며 장이 열린 지 불과 약 3분 만에 5천피(코스피 5,000)이 무너지고 서킷 브레이커가 발동됐다.

서킷브레이커가 해제된 후에는 연이어 프로그램 매도호가 일시효력정지(매도 사이드카)가 발동됐다.

오전 9시 53분 현재 코스피는 7.74% 하락한 4,967.05다. 삼성전자는 8.54% 내린 18만4천300원, SK하이닉스는 10.52% 내린 113만1천원에 거래되고 있다.

삼성전자와 SK하이닉스를 보유한 개인 투자자들은 전날에 이어 폭락세가 이어지자 당혹감을 넘어 망연자실한 분위기다.

코스콤 CHECK에 따르면 최근 한 달간(10월 6일∼11월 5일) 개인은 삼성전자를 5조8천323억원, SK하이닉스를 3조3천601억원 순매수했다. 상장지수펀드(ETF)를 포함한 국내 상장 종목 중 첫 번째, 두 번째로 많은 액수다.

코스피가 5.28% 내렸던 전날에도 삼성전자를 1조2천277억원, SK하이닉스는 8천902억원 순매수했다. 5일 삼성전자는 7.57% 내린 20만1천500원에, SK하이닉스는 9.46% 하락한 126만4천원에 각각 정규장 거래를 마쳤다.

"""
result = client.chat(model=MODEL, messages=[{"role":"user", "content":prompt}], format=NewsResult.model_json_schema())
print(result)

# 답변 형식 점검
response = NewsResult.model_validate_json(result.message.content)

print(response.summary)
print(response.sentiment)
print(response.keyword)
```

```python
import requests
from ollama import Client

OLLAMA_HOST = "http://localhost:11434"
MODEL = "qwen3.5:9b"

client = Client(host=OLLAMA_HOST)

def ask(messages, temperature=0):
    response = client.chat(
        model=MODEL,
        messages=messages,
        stream=False,
        think=False,
        options={"temperature": temperature},
    )
    return response.message.content


history = [{"role":"system", "content":"당신은 친절한 사내 도우미이다. 한국어로 3문장 이내로 답한다."}]

while True:
    q = input("나: ")
    if q in ("q", "종료"):
        break

    history.append({"role":"user", "content":q})
    print("AI: ", end="")
    answer = ""

    for chunk in client.chat(model=MODEL, messages=history, stream=True, think=False):
        content = chunk.message.content
        print(content, end="", flush=True)
        answer += content

    print()
    history.append({"role":"assistant", "content":answer})
```
