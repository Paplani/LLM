### Simple Prompt 버전
- input : 텍스트 전용 메세지 입력

```python
from openai import OpenAI

client = OpenAI()

response = client.responses.create(
    model=MODEL,
    input="Write a one-sentence bedtime story about a unicorn.",
    stream=True
)

print(response)
# 전체 내용 한꺼번에 받을 때
# print(response.output_text)

for event in response:
    print(event)
```

```python
from openai import OpenAI

client = OpenAI()

context = [
    {"role":"system", "content":"You are a helpful assistant."},
    {"role":"user", "content":"파이썬에서 리스트 컴프리헨션이 뭔지 3줄로 응답해줘."}
]

response = client.responses.create(
    model=MODEL,
    input=context,
)

# print(response)
print(response.output_text)
```

- instructions  : 모델의 전반적인 행동 방식이나 응답 원칙을 지정하는 지침(input에 전달하는 것보다 우선순위가 있음)

```python
response = client.responses.create(
    model=MODEL,
    instructions="Always answer in English",
    input=[{"role":"user", "content":"한국어로 답변해줘"}],
)
print(response.output_text)
```

```python
instructions = """
You are a backend API assistant.
- Answer concisely and accurately.
- If unsure, say you are unsure.
- Do not hallucinate APIs or facts
- Prefer bullet points over long prose.
"""


response = client.responses.create(
    model=MODEL,
    instructions=instructions,
    input="Are semicolons optional in JavaScript?"
)
print(response.output_text)
```

- reasoning : 추론 모델의 내부 강도 조절
    - low : 빠르게, 직관적으로
    - medium : 일반적인 설명, 추론 (기본값)
    - high : 복잡한 문제를 더 깊게 추론

```python
response = client.responses.create(
    model=MODEL,
    reasoning={"effort": "low"},
    input=[
        {"role": "developer", "content": "Talk like a pirate."},
        {"role": "user", "content": "Are semicolons optional in JavaScript?"},
    ],
)

print(response.output_text)
```

#### 멀티턴 대화
- 이전 대화 내용 유지하기
- stateless : LLM API는 요청을 독립적으로 처리하기 때문에 이전 대화 내용을 다음 요청에 전달해야 문맥을 이어나갈 수 있다.
- 유지시키는 방법
    1. previous_response_id 이용
    2. 개발자가 문맥 관ㄹ
    3. conversations API이용

```python
# 프랑스우 수도는 어디야? => 파리 => 그 곳의 인구는 얼마나 돼?

context = [{"role":"user", "content":"what is the capital of France?"}]

res1 = client.responses.create(
    model=MODEL,
    reasoning={"effort": "none"},
    input=context
)

print(res1.output_text)

context += [{"role":"user", "content":"what is its populations?"}]

res2 = client.responses.create(
    model=MODEL,
    reasoning={"effort": "none"},
    input=context,
    previous_response_id=res1.id
)

print(res2.output_text)
```

```python
context = [
    {"role":"user", "content":"what is the capital of France?"},
    {"role":"assistant", "content":"Paris"},
    {"role":"assistant", "content":"What is its populations?"}
    ]

res2 = client.responses.create(
    model=MODEL,
    reasoning={"effort": "none"},
    input=context,
)

print(res2.output_text)
```

```python
# 프랑스우 수도는 어디야? => 파리 => 그 곳의 인구는 얼마나 돼?

context = [{"role":"user", "content":"what is the capital of France?"}]

res1 = client.responses.create(
    model=MODEL,
    reasoning={"effort": "none"},
    input=context, 
    store=True      # Default
)

print(res1.output_text)

context += res1.output
context += [{"role":"user", "content":"what is its populations?"}]

res2 = client.responses.create(
    model=MODEL,
    reasoning={"effort": "none"},
    input=context,
)

print(res2.output_text)
```

- conversation API
    - 대화 객체로 관리
    - 여러 세션이나 작업에 걸쳐 대화를 이어나갈 수 있다.

```python
import openai

# conversation 객체 생성
conversation = openai.conversations.create()

res1 = client.responses.create(
    model=MODEL,
    reasoning={"effort": "none"},
    input=[{"role":"user", "content":"농담하나 해줘"}],
    conversation=conversation.id
)

res2 = client.responses.create(
    model=MODEL,
    reasoning={"effort": "none"},
    input=[{"role":"user", "content":"그게 왜 재밌다고 생각해?"}],
    conversation=conversation.id
)
print(res1.output_text)
print(res2.output_text)
```

#### 파라메터
- reasoning : none 인 경우에만 temperature 사용가능
- reasoning 모델에서는 reasoning none 안할 경우에 'temperature' is not supported with this model 에러뜸

```python
res1 = client.responses.create(
    model=MODEL,
    reasoning={"effort": "none"},
    input=[{"role":"user", "content":"Tell me a funny story."}],
    temperature=1.5
)

print(res1.output_text)
```
